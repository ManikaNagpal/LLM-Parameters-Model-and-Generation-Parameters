This code does the following tasks:
1) Loads a model so you can run LLM inference.

2) Defines generation parameters so you can control:

*   creativity
*   randomness
*   coherence
*   length
3)  Generates text using those parameters

It will allow you to experiment with different values and understand their effect.

# Importing Required Classes

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, TextGenerationPipeline
```

# Load Model and Tokenizer

```python
model_name = "meta-llama/Llama-2-7b-chat-hf"
```

```python
from huggingface_hub import notebook_login
notebook_login()  # Login with your HF token
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
model = AutoModelForCausalLM.from_pretrained(model_name, device_map="auto", torch_dtype="auto")
```

# Create a Pipeline

```python
pipe = TextGenerationPipeline(model=model, tokenizer=tokenizer)
```

# Declaring Generation Parameters

```python
generation_params = {
    "temperature": 0.7,     # Controls randomness
    "top_p": 0.9,           # Nucleus sampling (probability mass)
    "top_k": 50,            # Consider top-k likely tokens
    "max_new_tokens": 100,  # Maximum tokens to generate
    "repetition_penalty": 1.0, # Similar to frequency/presence penalty
}
```

# Define a Prompt

```python
prompt = "Explain the importance of data privacy in simple words."
```

# Generate Output

```python
output = pipe(
    prompt,
    **generation_params
)
```

```python
print(output[0]["generated_text"])
```
